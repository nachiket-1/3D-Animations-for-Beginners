## How is 3D Different from 2D?

In Part 1, we created axes like this:
```python
fig, ax = plt.subplots()          # standard 2D axes
```

For 3D, we add one line:
```python
fig = plt.figure()
ax  = fig.add_subplot(projection='3d')   # 3D axes!
```

Everything else — `FuncAnimation`, `set_data()`, `blit` — works almost the same way.

**The new superpower: `ax.view_init(elev, azim)`**

```
elev  = elevation angle (how high up you're looking from)
azim  = azimuth angle   (rotating horizontally around the scene)
```

By changing `azim` every frame, we get **auto-rotation for free** — no physics needed!

```python
def update(frame):
    ax.view_init(elev=20, azim=frame * 2)   # rotate 2 degrees per frame
```

<div style="background: linear-gradient(90deg, #7f5af0 0%, #2cb67d 100%); padding: 16px 24px; border-radius: 12px; margin: 16px 0;">
  <h2 style="color: white; margin: 0; font-size: 1.5em; text-shadow: 0 1px 6px rgba(0,0,0,0.4);">Animation 1 — Rotating 3D Scatter Plot</h2>
</div>

### The idea

We'll create a cloud of 3D points — colored by their Z height — and **slowly rotate the camera** around them. The data stays still; the *viewpoint* moves.

This is the easiest possible 3D animation and the pattern used in almost every rotating 3D demo you've seen!

```
Frame 0:  azim =   0°  → looking from the front
Frame 45: azim =  90°  → looking from the right
Frame 90: azim = 180°  → looking from the back
Frame 180: azim = 360° → full rotation complete
```

````python
# ============================================================
# ANIMATION 1: Rotating 3D Scatter Cloud
# ============================================================
np.random.seed(42)
N = 200   # number of points

# Random 3D points in a sphere-ish shape
theta  = np.random.uniform(0, 2*np.pi, N)
phi    = np.random.uniform(0, np.pi, N)
r      = np.random.uniform(0.3, 1.0, N)
xs     = r * np.sin(phi) * np.cos(theta)
ys     = r * np.sin(phi) * np.sin(theta)
zs     = r * np.cos(phi)

fig = plt.figure(figsize=(8, 7))
ax  = fig.add_subplot(projection='3d')
ax.set_facecolor('#0d1117')
fig.patch.set_facecolor('#0d1117')

# Color each point by its Z value
colors = cm.plasma((zs + 1) / 2)   # normalise z from [-1,1] → [0,1]

scatter = ax.scatter(xs, ys, zs,
                     c=colors, s=18, alpha=0.85,
                     edgecolors='none', depthshade=True)

ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.set_zlim(-1.1, 1.1)
ax.set_title('Rotating 3D Point Cloud', color='white',
             fontsize=14, pad=15)

# Remove grey panes for a cleaner look
ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.xaxis.pane.set_edgecolor('#222')
ax.yaxis.pane.set_edgecolor('#222')
ax.zaxis.pane.set_edgecolor('#222')
ax.tick_params(colors='#555')

# KEY TRICK: just change the camera angle each frame!
def update(frame):
    ax.view_init(elev=20, azim=frame * 2)   # 2 degrees per frame
    return scatter,

ani = animation.FuncAnimation(
    fig, update,
    frames=180,      # 180 frames × 2° = full 360° rotation
    interval=40,
    blit=False       # must be False for 3D
)

plt.tight_layout()
HTML(ani.to_jshtml())

# TRY: Change elev=20 to elev=60 for a top-down view
# TRY: Change cm.plasma to cm.viridis or cm.cool
````

### What just happened?

| Line | What it does |
|------|--------------|
| `projection='3d'` | Switches the axes to 3D mode |
| `ax.scatter(x, y, z)` | Plots points in 3D space |
| `depthshade=True` | Points further away appear dimmer (depth effect) |
| `ax.view_init(elev, azim)` | Moves the camera — **this is the whole animation** |
| `blit=False` | 3D axes don't support blit — always use False |

> Notice `blit=False` — this is a **key difference from Part 1**. 3D axes redraw entirely each frame.

### The idea

A **surface plot** fills in a full 3D mesh — like a terrain map or ocean surface. We'll animate it as a **rippling wave** by updating the Z values each frame.

**New concept: Meshgrid**

To draw a surface, we need X, Y, Z at every grid point:
```python
x = np.linspace(-5, 5, 50)     # 50 x-values
y = np.linspace(-5, 5, 50)     # 50 y-values
X, Y = np.meshgrid(x, y)       # 50×50 grid of (x,y) pairs
Z = np.sin(X) * np.cos(Y)      # Z value at each grid point
```

To animate it, we shift Z over time:
```python
Z = np.sin(X - t) * np.cos(Y - t)   # wave moves as t increases
```

````python
# ============================================================
# ANIMATION 2: Animated 3D Surface Wave
# ============================================================

# Build the grid
x  = np.linspace(-4, 4, 60)
y  = np.linspace(-4, 4, 60)
X, Y = np.meshgrid(x, y)

fig = plt.figure(figsize=(9, 7))
ax  = fig.add_subplot(projection='3d')
fig.patch.set_facecolor('#0d1117')
ax.set_facecolor('#0d1117')

# Clean up panes
for pane in [ax.xaxis.pane, ax.yaxis.pane, ax.zaxis.pane]:
    pane.fill = False
    pane.set_edgecolor('#1c1c2e')
ax.tick_params(colors='#555')

ax.set_zlim(-2, 2)
ax.set_title('Rippling 3D Surface Wave', color='white',
             fontsize=14, pad=15)

# Draw the initial surface
Z    = np.sin(np.sqrt(X**2 + Y**2))   # circular ripple pattern
surf = [ax.plot_surface(X, Y, Z,
                        cmap='plasma',
                        linewidth=0,
                        antialiased=True,
                        alpha=0.9)]

# For surfaces we REMOVE the old one and draw a new one each frame
# (unlike lines where we use set_data)
def update(frame):
    surf[0].remove()                   # remove previous surface
    t = frame * 0.15                   # time offset
    Z = np.sin(np.sqrt(X**2 + Y**2) - t) * np.cos(0.5 * t)
    surf[0] = ax.plot_surface(X, Y, Z,
                              cmap='plasma',
                              linewidth=0,
                              antialiased=True,
                              alpha=0.9)
    ax.view_init(elev=28, azim=frame * 1.2)   # slow rotation too!
    return surf[0],

ani = animation.FuncAnimation(
    fig, update,
    frames=120,
    interval=50,
    blit=False
)

plt.tight_layout()
HTML(ani.to_jshtml())

# TRY: Change cmap='plasma' to cmap='ocean' or cmap='hot'
# TRY: Replace np.sin(np.sqrt(X**2 + Y**2) - t) with np.sin(X - t) * np.cos(Y)
````

### Key pattern: Remove & Redraw

For surfaces, we can't use `set_data()`. Instead:

```python
surf = [ax.plot_surface(...)]   # store in a list (mutable!)

def update(frame):
    surf[0].remove()            # delete old surface
    surf[0] = ax.plot_surface() # draw new one
```

We use a **list** (`surf = [...]`) instead of a plain variable because Python functions can't reassign outer variables — but they *can* modify a list's contents.

> This is the standard pattern for animating any filled 3D object — surfaces, filled contours, 3D bars, etc.

### The idea

A **helix** is a 3D spiral — think of a spring, a DNA strand, or a spiral staircase. We'll animate it *growing* from nothing to a full double helix, while the camera rotates around it.

**The math of a helix:**
```
x = cos(t)
y = sin(t)
z = t          ← this is what makes it spiral upward!
```

For a **double helix** (like DNA), we just offset the second strand by π:
```
Strand 1: x = cos(t),     y = sin(t),     z = t
Strand 2: x = cos(t + π), y = sin(t + π), z = t
```

````python
# ============================================================
# ANIMATION 3: Growing Double Helix
# ============================================================

# Full helix data
t_full  = np.linspace(0, 6 * np.pi, 300)   # 3 full turns
x1 = np.cos(t_full);   y1 = np.sin(t_full);  z1 = t_full / (2*np.pi)
x2 = np.cos(t_full + np.pi); y2 = np.sin(t_full + np.pi); z2 = z1

fig = plt.figure(figsize=(7, 9))
ax  = fig.add_subplot(projection='3d')
fig.patch.set_facecolor('#0a0a0f')
ax.set_facecolor('#0a0a0f')

for pane in [ax.xaxis.pane, ax.yaxis.pane, ax.zaxis.pane]:
    pane.fill = False
    pane.set_edgecolor('#111')
ax.tick_params(colors='#444')
ax.set_axis_off()   # clean look — no axes

# Initial empty lines for both strands
line1, = ax.plot([], [], [], color='#f72585', linewidth=2.5, alpha=0.9)
line2, = ax.plot([], [], [], color='#4cc9f0', linewidth=2.5, alpha=0.9)

# Rungs connecting the two strands (like DNA base pairs)
rung_lines = []
rung_indices = range(0, len(t_full), 15)   # every 15th point
for i in rung_indices:
    rung, = ax.plot([], [], [], color='#ffffff', linewidth=0.8, alpha=0.3)
    rung_lines.append((i, rung))

# Glowing dot at the growing tip
tip1, = ax.plot([], [], [], 'o', color='#f72585', markersize=8)
tip2, = ax.plot([], [], [], 'o', color='#4cc9f0', markersize=8)

title_text = ax.text2D(0.5, 0.97, 'Growing Double Helix',
                        transform=ax.transAxes,
                        ha='center', color='white',
                        fontsize=13, fontweight='bold')

ax.set_xlim(-1.5, 1.5)
ax.set_ylim(-1.5, 1.5)
ax.set_zlim(0, 3)

def update(frame):
    n = frame + 1                        # how many points to show

    # Update strand lines
    line1.set_data(x1[:n], y1[:n])
    line1.set_3d_properties(z1[:n])

    line2.set_data(x2[:n], y2[:n])
    line2.set_3d_properties(z2[:n])

    # Moving tip dots
    tip1.set_data([x1[n-1]], [y1[n-1]])
    tip1.set_3d_properties([z1[n-1]])
    tip2.set_data([x2[n-1]], [y2[n-1]])
    tip2.set_3d_properties([z2[n-1]])

    # Draw rungs only where both strands have grown
    for idx, rung in rung_lines:
        if idx < n:
            rung.set_data([x1[idx], x2[idx]], [y1[idx], y2[idx]])
            rung.set_3d_properties([z1[idx], z2[idx]])
        else:
            rung.set_data([], [])
            rung.set_3d_properties([])

    ax.view_init(elev=15, azim=frame * 1.5)   # rotate as helix grows

ani = animation.FuncAnimation(
    fig, update,
    frames=len(t_full),
    interval=20,
    blit=False
)

plt.tight_layout()
HTML(ani.to_jshtml())

# TRY: Change range(0, len(t_full), 15) to range(0, len(t_full), 8) for more rungs
# TRY: Change 6*np.pi to 10*np.pi for more turns
````

### New 3D concept: `set_3d_properties()`

In 2D we used:
```python
line.set_data(x, y)
```

In 3D we need to set Z separately:
```python
line.set_data(x, y)           # sets X and Y
line.set_3d_properties(z)     # sets Z
```

This is because `set_data()` was designed for 2D and the 3D toolkit extends it with this extra method. Always use both together for 3D lines!

  <h2 style="color: #1a1200; margin: 0; font-size: 1.5em;">Animation 4 — Orbiting Moon Around a Planet</h2>
</div>

### The idea

We combine everything from this notebook into one final animation:
- A **3D sphere** (the planet) drawn with `plot_surface`
- A **smaller sphere** (the moon) orbiting around it
- An **orbit trail** drawn as a 3D line
- A **slow camera rotation** around the whole scene

**How to draw a sphere:**
```python
u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 50)
x = radius * np.outer(np.cos(u), np.sin(v))
y = radius * np.outer(np.sin(u), np.sin(v))
z = radius * np.outer(np.ones(50), np.cos(v))
ax.plot_surface(x, y, z)
```

````python
# ============================================================
# ANIMATION 4: Planet + Orbiting Moon
# ============================================================

fig = plt.figure(figsize=(9, 8))
ax  = fig.add_subplot(projection='3d')
fig.patch.set_facecolor('#020409')
ax.set_facecolor('#020409')

for pane in [ax.xaxis.pane, ax.yaxis.pane, ax.zaxis.pane]:
    pane.fill = False
    pane.set_edgecolor('#050a10')
ax.set_axis_off()

# Starfield
np.random.seed(7)
sx = np.random.uniform(-4, 4, 300)
sy = np.random.uniform(-4, 4, 300)
sz = np.random.uniform(-4, 4, 300)
ax.scatter(sx, sy, sz, s=np.random.uniform(0.5, 3, 300),
           color='white', alpha=0.25, zorder=0)

# ── Draw the planet (static sphere)
u = np.linspace(0, 2*np.pi, 60)
v = np.linspace(0, np.pi,   60)
px = 1.0 * np.outer(np.cos(u), np.sin(v))
py = 1.0 * np.outer(np.sin(u), np.sin(v))
pz = 1.0 * np.outer(np.ones(60), np.cos(v))
ax.plot_surface(px, py, pz, cmap='Blues',
                alpha=0.9, linewidth=0, zorder=1)

# Orbit path (faint ring)
ORBIT_R = 2.2
ORBIT_TILT = 15   # degrees
tilt_rad = np.radians(ORBIT_TILT)
orbit_t = np.linspace(0, 2*np.pi, 200)
orbit_x = ORBIT_R * np.cos(orbit_t)
orbit_y = ORBIT_R * np.sin(orbit_t) * np.cos(tilt_rad)
orbit_z = ORBIT_R * np.sin(orbit_t) * np.sin(tilt_rad)
ax.plot(orbit_x, orbit_y, orbit_z,
        color='#ffffff', alpha=0.12, linewidth=1)

# Moon — small sphere, stored in a list for remove/redraw
um = np.linspace(0, 2*np.pi, 25)
vm = np.linspace(0, np.pi,   25)
moon_surf = [None]

# Moon trail line
trail_line, = ax.plot([], [], [], color='#ffd166',
                       linewidth=1, alpha=0.5)
trail_x, trail_y, trail_z = [], [], []
TRAIL = 40   # trail length in frames

# Glow dot at moon center
moon_dot, = ax.plot([], [], [], 'o',
                     color='#ffd166', markersize=10, alpha=0.9)

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_zlim(-3, 3)

title = ax.text2D(0.5, 0.96, 'Planet & Orbiting Moon',
                   transform=ax.transAxes, ha='center',
                   color='white', fontsize=13, fontweight='bold')

def update(frame):
    angle = frame * (2 * np.pi / 120)   # full orbit in 120 frames

    # Moon center position
    mx = ORBIT_R * np.cos(angle)
    my = ORBIT_R * np.sin(angle) * np.cos(tilt_rad)
    mz = ORBIT_R * np.sin(angle) * np.sin(tilt_rad)

    # Remove old moon surface and draw new one at new position
    if moon_surf[0] is not None:
        moon_surf[0].remove()
    mr = 0.28
    msx = mr * np.outer(np.cos(um), np.sin(vm)) + mx
    msy = mr * np.outer(np.sin(um), np.sin(vm)) + my
    msz = mr * np.outer(np.ones(25), np.cos(vm)) + mz
    moon_surf[0] = ax.plot_surface(msx, msy, msz,
                                    color='#c0c0c0',
                                    alpha=0.95, linewidth=0)

    # Update trail
    trail_x.append(mx); trail_y.append(my); trail_z.append(mz)
    if len(trail_x) > TRAIL:
        trail_x.pop(0); trail_y.pop(0); trail_z.pop(0)
    trail_line.set_data(trail_x, trail_y)
    trail_line.set_3d_properties(trail_z)

    # Glow dot
    moon_dot.set_data([mx], [my])
    moon_dot.set_3d_properties([mz])

    # Slow camera rotation
    ax.view_init(elev=25, azim=frame * 1.0)

ani = animation.FuncAnimation(
    fig, update,
    frames=240,      # 2 full orbits
    interval=35,
    blit=False
)

plt.tight_layout()
HTML(ani.to_jshtml())

# TRY: Change ORBIT_TILT = 15 to 45 for a tilted orbit
# TRY: Change ORBIT_R = 2.2 to 1.6 for a closer moon
# TRY: Add a second moon with a different orbit radius!
````
